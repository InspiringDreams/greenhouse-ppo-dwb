# Figure 4.4: time to goal and Mann-Whitney test

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

BASE = str(Path.home() / 'greenhouse_ws')
t = pd.read_csv(f'{BASE}/runs/ttg_times.csv')
t = t[~((t['planner'].str.upper() == 'PPO') & (t['world'] == 'greenhouse_low_medium_s26'))]
ppo = t[t['planner'].str.upper() == 'PPO']['seconds'].values
dwb = t[t['planner'].str.upper() == 'DWB']['seconds'].values
print(len(ppo), len(dwb), np.median(ppo), np.median(dwb))

rng = np.random.default_rng(1)

fig, ax = plt.subplots(figsize=(6, 4.5))
ax.boxplot([ppo, dwb], widths=0.5, showfliers=False)
ax.scatter(1 + rng.uniform(-0.18, 0.18, len(ppo)), ppo, s=12, alpha=0.4, color='#2a78d6')
ax.scatter(2 + rng.uniform(-0.18, 0.18, len(dwb)), dwb, s=12, alpha=0.4, color='#eb6834')
ax.text(1.3, np.median(ppo), f'median {np.median(ppo):.1f} s', va='center')
ax.text(2.3, np.median(dwb), f'median {np.median(dwb):.1f} s', va='center')
ax.set_yscale('log')
ax.minorticks_off()
ax.set_yticks([40, 50, 60, 80, 100, 150, 200])
ax.set_yticklabels(['40', '50', '60', '80', '100', '150', '200'])
ax.set_xticks([1, 2])
ax.set_xticklabels(['PPO', 'DWB'])
ax.set_xlim(0.5, 3.0)
ax.set_ylabel('Time to goal (seconds, log scale)')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('fig4_4_time.png', dpi=300)
```
