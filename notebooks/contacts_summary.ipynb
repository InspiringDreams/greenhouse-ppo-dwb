# Contact counts and durations

```python
from pathlib import Path
import pandas as pd

BASE = str(Path.home() / 'greenhouse_ws')
d = pd.read_csv(f'{BASE}/runs/episodes_dwb_v2.csv')
c = pd.read_csv(f'{BASE}/scripts/true_contacts.csv')
c = c[c.planner == 'DWB'][['world', 'true_contact_samples']]
m = d.merge(c, on='world')
m['contact'] = m.true_contact_samples > 0
print(len(d), len(m))
print(m.groupby('contact')['recovery_count'].agg(['count', 'mean', 'median', 'sum']))
print('episodes with at least 1 recovery:', (m.recovery_count > 0).sum())
```

```python
c = pd.read_csv(f'{BASE}/scripts/true_contacts.csv')
c = c[c.true_contact_samples > 0]
c['seconds'] = c.true_contact_samples * 0.12
print(c.groupby('planner')['seconds'].agg(['count', 'median', 'max']))
```
