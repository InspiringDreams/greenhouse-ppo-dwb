# Figure 4.1: success rate at each tolerance

```python
import pandas as pd
ppo = pd.read_csv('episodes_ppo.csv')
dwb = pd.read_csv('episodes_dwb_v2.csv')
print(len(ppo), len(dwb))   # should print 270 270
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ppo_dist = ppo['truth_dist_m']
dwb_dist = dwb['truth_min_dist_m']
```

```python
tolerances = [0.25, 0.50, 0.75, 1.00]
ppo_rate = [(ppo_dist < t).mean() * 100 for t in tolerances]
dwb_rate = [(dwb_dist < t).mean() * 100 for t in tolerances]
```

```python
plt.figure(figsize=(7, 4))
plt.plot(tolerances, ppo_rate, marker='o', color='#2a78d6', label='PPO')
plt.plot(tolerances, dwb_rate, marker='s', color='#eb6834', label='DWB')

for t, p, d in zip(tolerances, ppo_rate, dwb_rate):
    plt.text(t, p + 3, f'{p:.1f}%', ha='center')
    plt.text(t, d - 7, f'{d:.1f}%', ha='center')

plt.xticks(tolerances)
plt.ylim(0, 110)
plt.xlabel('Arrival tolerance (m)')
plt.ylabel('Episodes reaching goal (%)')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('fig4_1_success.png', dpi=300)
plt.show()
```
