# Figure 4.6: success by curvature and obstacle density

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE = str(Path.home() / 'greenhouse_ws')
ppo = pd.read_csv(f'{BASE}/runs/episodes_ppo.csv')
dwb = pd.read_csv(f'{BASE}/runs/episodes_dwb_v2.csv')

ppo['reached'] = ppo['truth_dist_m'].fillna(99) < 0.75
dwb['reached'] = dwb['truth_min_dist_m'] < 0.75

levels = [
    ('curvature', ['straight', 'low', 'high'], 'Row curvature'),
    ('density', ['sparse', 'medium', 'cluttered'], 'Obstacle density'),
]

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
for ax, (column, order, label) in zip(axes, levels):
    ppo_rate = [ppo[ppo[column] == lv]['reached'].mean() * 100 for lv in order]
    dwb_rate = [dwb[dwb[column] == lv]['reached'].mean() * 100 for lv in order]
    ax.plot(order, ppo_rate, marker='o', color='#2a78d6', label='PPO')
    ax.plot(order, dwb_rate, marker='s', color='#eb6834', label='DWB')
    for i, (a, b) in enumerate(zip(ppo_rate, dwb_rate)):
        ax.text(i, a + 3, f'{a:.1f}%', ha='center', color='#2a78d6')
        ax.text(i, b - 8, f'{b:.1f}%', ha='center', color='#eb6834')
    ax.set_xlim(-0.4, 2.4)
    ax.set_xlabel(label)
    ax.set_ylim(40, 110)
    ax.grid(axis='y', alpha=0.3)
axes[0].set_ylabel('Episodes reaching goal within 0.75 m (%)')
axes[0].legend(loc='lower left')
plt.tight_layout()
plt.savefig('fig4_6_curvature_density.png', dpi=300)
```
