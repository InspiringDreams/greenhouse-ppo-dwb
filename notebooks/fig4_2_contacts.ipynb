# Figure 4.2: contacts

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE = str(Path.home() / 'greenhouse_ws')
data = pd.read_csv(f'{BASE}/scripts/true_contacts.csv')
data = data[data['true_contact_samples'] > 0]
data['seconds'] = data['true_contact_samples'] * 0.12

ppo = data[data['planner'] == 'PPO']['seconds']
dwb = data[data['planner'] == 'DWB']['seconds']

plt.figure(figsize=(7, 3))
plt.scatter(ppo, [1] * len(ppo), s=70, alpha=0.6, color='#2a78d6', marker='o', label=f'PPO ({len(ppo)} episodes)')
plt.scatter(dwb, [0] * len(dwb), s=70, alpha=0.6, color='#eb6834', marker='s', label=f'DWB ({len(dwb)} episodes)')

plt.xscale('log')
plt.xticks([0.25, 0.5, 1, 2, 5, 10, 20, 60], ['0.25', '0.5', '1', '2', '5', '10', '20', '60'])
plt.yticks([0, 1], ['DWB', 'PPO'])
plt.ylim(-0.6, 1.6)
plt.xlabel('Time in contact with an object (seconds, log scale)')
plt.legend(loc='lower right')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('fig4_2_contacts.png', dpi=300)
```
